```scala
%spark.pyspark
from pyspark.sql import functions as F
from pyspark.sql.window import Window

users = spark.createDataFrame(
    [("u1", "Berlin"), ("u2", "Berlin"), ("u3", "Munich"), ("u4", "Hamburg")],
    ["user_id", "city"]
)

orders = spark.createDataFrame(
    [
        ("o1", "u1", "p1", 2, 10.0),
        ("o2", "u1", "p2", 1, 30.0),
        ("o3", "u2", "p1", 1, 10.0),
        ("o4", "u2", "p3", 5, 7.0),
        ("o5", "u3", "p2", 3, 30.0),
        ("o6", "u3", "p3", 1, 7.0),
        ("o7", "u4", "p1", 10, 10.0)
    ],
    ["order_id", "user_id", "product_id", "qty", "price"]
)

products = spark.createDataFrame(
    [("p1", "Ring VOLA"), ("p2", "Ring POROG"), ("p3", "Ring TISHINA")],
    ["product_id", "product_name"]
)
```

```scala
%spark.pyspark

orders_with_rev = orders.withColumn("revenue", F.col("qty") * F.col("price"))

enriched_orders = orders_with_rev \
    .join(users, on="user_id", how="left") \
    .join(products, on="product_id", how="left")

agg_df = enriched_orders.groupBy("city", "product_id", "product_name").agg(
    F.count("order_id").alias("orders_cnt"),
    F.sum("qty").alias("qty_sum"),
    F.sum("revenue").alias("revenue_sum")
)

window_spec = Window.partitionBy("city").orderBy(F.col("revenue_sum").desc())

mart_city_top_products = agg_df \
    .withColumn("rank", F.rank().over(window_spec)) \
    .filter(F.col("rank") <= 2) \
    .drop("rank") \
    .orderBy("city", F.col("revenue_sum").desc())
```

```scala
%spark.pyspark

hdfs_path = "hdfs:///tmp/sandbox_zeppelin/mart_city_top_products/"
s3_path = "s3a://hse-1/tmp/sandbox_zeppelin/mart_city_top_products/" 

mart_city_top_products.write.mode("overwrite").parquet(hdfs_path)

mart_city_top_products.write.mode("overwrite").parquet(s3_path)

result_df = spark.read.parquet(hdfs_path)

print("Итоговая витрина mart_city_top_products:")
result_df.show(truncate=False)
```
